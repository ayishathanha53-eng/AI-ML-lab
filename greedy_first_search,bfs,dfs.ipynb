```python
import heapq

# Define a simple graph using an adjacency list with edge weights
# Each value in the adjacency list is a list of tuples: (neighbor_node, weight)
graph = {
    'A': [('B', 1), ('C', 5)],
    'B': [('A', 1), ('D', 2), ('E', 2)],
    'C': [('A', 5), ('F', 1)],
    'D': [('B', 2)],
    'E': [('B', 2), ('F', 3)],
    'F': [('C', 1), ('E', 3)]
}

# Define heuristic values for Greedy Best-First Search (example values)
# These values represent the estimated cost from each node to the goal node (e.g., 'F')
heuristics = {
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 2,
    'E': 1,
    'F': 0  # Goal node has a heuristic of 0
}

#greedy first search

import heapq

def greedy_best_first_search_with_cost(graph, heuristics, start_node, goal_node):
    # Priority queue stores tuples: (heuristic_value, accumulated_cost, current_node, current_path)
    priority_queue = [(heuristics[start_node], 0, start_node, [start_node])]
    visited = set() # To keep track of visited nodes

    while priority_queue:
        heuristic_value, current_cost, current_node, current_path = heapq.heappop(priority_queue)

        if current_node == goal_node:
            return current_path, current_cost

        if current_node not in visited:
            visited.add(current_node)

            for neighbor_node, edge_weight in graph[current_node]:
                if neighbor_node not in visited:
                    new_cost = current_cost + edge_weight
                    heapq.heappush(priority_queue, (heuristics[neighbor_node], new_cost, neighbor_node, current_path + [neighbor_node]))
    return None, None # Goal not reachable, return None for path and cost

path_gb_cost, total_cost_gb = greedy_best_first_search_with_cost(graph, heuristics, 'A', 'F')

if path_gb_cost:
    print(f"Greedy Best-First Search Path from A to F: {path_gb_cost}")
    print(f"Total Cost: {total_cost_gb}")
else:
    print("Goal not reachable by Greedy Best-First Search from A to F.")
```

```python
from collections import deque

 # bfs
def bfs(graph, start_node, goal_node):
    visited = set()  # To keep track of visited nodes
    queue = deque([start_node])  # Initialize queue with the starting node
    path = {start_node: [start_node]}  # To reconstruct the path

    while queue:
        current_node = queue.popleft()

        if current_node == goal_node:
            return path[current_node]

        if current_node not in visited:
            visited.add(current_node)

            # Extract only the neighbor_node from the (neighbor_node, weight) tuple
            for neighbor_tuple in graph[current_node]:
                neighbor = neighbor_tuple[0] # Get the node name
                if neighbor not in visited:
                    queue.append(neighbor)
                    path[neighbor] = path[current_node] + [neighbor]
    return None # Goal not reachable

print("BFS path from A to F:", bfs(graph, 'A', 'F'))
```

```python
#dfs

def dfs(graph, start_node, goal_node):
    visited = set()  # To keep track of visited nodes
    stack = [(start_node, [start_node])]  # Stack stores (current_node, current_path)

    while stack:
        current_node, current_path = stack.pop()

        if current_node == goal_node:
            return current_path

        if current_node not in visited:
            visited.add(current_node)

            # Explore neighbors in reverse order to simulate standard DFS (leftmost child first)
            # Extract only the neighbor_node from the (neighbor_node, weight) tuple
            for neighbor_tuple in reversed(graph[current_node]):
                neighbor = neighbor_tuple[0] # Get the node name
                if neighbor not in visited:
                    stack.append((neighbor, current_path + [neighbor]))
    return None # Goal not reachable

print("DFS path from A to F:", dfs(graph, 'A', 'F'))
```
